# Test Linear Model

Test the LinearModel on electricity demand prediction.

## Workflow
1. Load and prepare training/test data.
2. Train LinearModel with RMSE loss (baseline).
3. Train LinearModel with pinball loss (quantile regression).
4. Generate and save predictions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
import os

sys.path.append('../Models')
from Linear import LinearModel
```

```python
# Load training and test datasets.
Data_train = pd.read_csv("../Data/train.csv", parse_dates=["Date"])
Data_test = pd.read_csv("../Data/test.csv", parse_dates=["Date"])
```

```python
def find_continuous_columns(X):
    """Identify non-binary columns (continuous features)."""
    continuous_cols = []
    for col in X.columns:
        s = X[col]
        if s.dtype == bool:
            continue
        vals = pd.unique(s.dropna())
        if len(vals) <= 2 and set(vals).issubset({0, 1}):
            continue
        continuous_cols.append(col)
    return continuous_cols


def normalize(X, scale_cols=None):
    """Standardize continuous features using mean and standard deviation.
    
    Parameters
    ----------
    X : pandas.DataFrame
        Feature matrix.
    scale_cols : list or None
        Columns to scale. If None, auto-detect continuous columns.
    
    Returns
    -------
    X_norm : pandas.DataFrame
        Normalized feature matrix.
    mean : pandas.Series
        Mean values used for scaling.
    std : pandas.Series
        Std values used for scaling.
    """
    X_norm = X.copy()
    if scale_cols is None:
        scale_cols = find_continuous_columns(X_norm)
    
    mean = X_norm[scale_cols].mean(axis=0)
    std = X_norm[scale_cols].std(axis=0)
    std = std.replace(0, 1.0)
    
    if len(scale_cols) > 0:
        X_norm[scale_cols] = (X_norm[scale_cols] - mean) / std
    
    return X_norm, mean, std


def submit_predictions(ids, predictions, filename="submission"):
    """Save predictions to CSV submission file.
    
    Parameters
    ----------
    ids : array_like
        Test sample IDs.
    predictions : array_like
        Predicted values.
    filename : str
        Output filename (without .csv extension).
    """
    submission_df = pd.DataFrame({
        "Id": ids,
        "Net_demand": predictions
    })
    
    filepath = f"../Results/{filename}.csv"
    submission_df.to_csv(filepath, index=False)
    print(f"Saved predictions to {filepath}")
```

## Data Preparation

```python
# Create feature matrix from training data.
df = Data_train.copy()

# Mark COVID lockdown period (structural break starting 2020-03-17).
df['Date'] = pd.to_datetime(df['Date'])
df['Covid_Flag'] = (df['Date'] >= '2020-03-17').astype(int)

# Temperature in Kelvin: convert to heating/cooling degree days.
kelvin_0 = 273.15
seuil_chauffage = kelvin_0 + 15  # ~15°C heating threshold
seuil_clim = kelvin_0 + 22       # ~22°C cooling threshold

df['Heating_Degree'] = df['Temp'].apply(lambda x: max(0, seuil_chauffage - x))
df['Cooling_Degree'] = df['Temp'].apply(lambda x: max(0, x - seuil_clim))

# Select features (drop target and identifiers).
cols_to_drop = ["Net_demand", "Date", "Solar_power", "Wind_power", "Load", 
                "Nebulosity", "Nebulosity_weighted", "Id"]
cols_to_drop = [c for c in df.columns if c in cols_to_drop]

X_train = df.drop(columns=cols_to_drop)
y_train = df["Net_demand"]

X_test = Data_test.drop(columns=["Date", "Usage", "Id"])

# One-hot encode day-of-week.
X_train = pd.get_dummies(X_train, columns=['WeekDays'], prefix='WeekDays', drop_first=True, dtype=float)
X_test = pd.get_dummies(X_test, columns=['WeekDays'], prefix='WeekDays', drop_first=True, dtype=float)

# Align test columns with train.
X_test = X_test.reindex(columns=X_train.columns, fill_value=0.0)

y_train_np = y_train.values
```

```python
# Normalize continuous features.
scale_cols = find_continuous_columns(X_train)
X_train, X_mean, X_std = normalize(X_train, scale_cols=scale_cols)
X_test, _, _ = normalize(X_test, scale_cols=scale_cols)

# Convert to numpy for model.
X_train_np = X_train.values
X_test_np = X_test.values
```

```python
# Train baseline model with RMSE loss.
model_rmse = LinearModel(learning_rate=0.02, maxIter=10000)
model_rmse.fit(X_train_np, y_train_np, loss="rmse")
y_test_pred_rmse = model_rmse.predict(X_test_np)

print("RMSE model trained.")
```

```python
# Train pinball/quantile regression model (tau=0.8).
tau = 0.8
model_pinball = LinearModel(learning_rate=0.02, maxIter=17000, tau=tau)
model_pinball.fit(X_train_np, y_train_np, loss="pinball", verbose=False)
y_test_pred_pinball = model_pinball.predict(X_test_np)
```

```python
# Analyze feature importance from pinball model.
feature_names = list(X_train.columns)
coefs = np.asarray(model_pinball.weights).reshape(-1)

# Mark binary vs continuous features.
is_binary = []
for c in feature_names:
    vals = pd.unique(X_train[c].dropna())
    is_binary.append(len(vals) <= 2 and set(vals).issubset({0, 1}))

rank_df = pd.DataFrame({
    "feature": feature_names,
    "coef": coefs,
    "abs_coef": np.abs(coefs),
    "type": np.where(is_binary, "binary/one-hot", "continuous")
}).sort_values("abs_coef", ascending=False)

print("Top 20 features by importance:")
print(rank_df.head(20))
```

## Save Predictions

```python
# Save both models' predictions.
submit_predictions(Data_test["Id"], y_test_pred_rmse, filename="linear_rmse_submission")
submit_predictions(Data_test["Id"], y_test_pred_pinball, filename="linear_pinball_submission")
```
